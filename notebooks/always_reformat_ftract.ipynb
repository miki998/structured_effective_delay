## Description

Notebook to quickly and dirtily reformat the F-Tract, cause otherwise unusable.

```python
import gzip

# Combine all the txt files into arrays with structured names
scales = ["scale1", "scale2", "scale3", "scale4"]
age_ranges = ["0_15", "0_inf", "15_inf", "18_60"]
delays = ["50", "100", "200", "400", "800"]
features = ["feature_peak_delay_zth5", "feature_dcm_axonal_delay"]


agg_dict = {}

for scale in scales:
    for age_range in age_ranges:
        for delay_max in delays:
            for feature in features:
                curpath = f"/Users/mikichan/Desktop/mywork/data_resources/atlas_data/fiber_atlas_2018/Lausanne2018_FTRACT_NEW/Lausanne2018-{scale}/{age_range}/{delay_max}/max_peak_delay_{delay_max}__zth5/min_value_gen__0/{feature}/"

                if op.exists(op.join(curpath, 'nanquantile_0.5.txt.gz')):
                    with gzip.open(op.join(curpath, 'nanquantile_0.5.txt.gz'), 'rb') as f_gz, open(op.join(curpath, 'nanquantile_0.5.txt'), 'wb') as f_out:
                        f_out.write(f_gz.read())
                with open(op.join(curpath, 'nanquantile_0.5.txt')) as f:
                    text = f.readlines()

                y_ground_mat = []
                for t in text[8:]:
                # for t in text:
                    y_ground_mat.append(t.split(' '))
                y_ground_mat = np.array(y_ground_mat).astype(float)
                y_ground_mat = np.nan_to_num(y_ground_mat)

                key_name = f"{scale}__{age_range}__{delay_max}__{feature}"
                agg_dict[key_name] = y_ground_mat
```
